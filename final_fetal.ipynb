```python
from sklearn.preprocessing import StandardScaler,MinMaxScaler,RobustScaler
from sklearn.model_selection import train_test_split,StratifiedKFold,cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, f1_score,recall_score,precision_score,classification_report,confusion_matrix
```

```python
std = StandardScaler()
minmax = MinMaxScaler()
R_scale= RobustScaler()
log = LogisticRegression(random_state = 0, max_iter=1000, class_weight="balanced")
knn = KNeighborsClassifier(n_neighbors=5)
stf = StratifiedKFold(n_splits =7,random_state=42,shuffle=True)
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size = 0.2,shuffle=True,random_state=42,stratify = y)
X_train.shape,X_test.shape
```

```python
pipeline1 = Pipeline([
    ("StandardScaler",std),
    ("Model_1",log)
])
pipeline2 = Pipeline([
    ("MinMax", minmax),
    ("Model_1",log)
])
pipeline3 = Pipeline([
    ("RobuseScaler", R_scale),
    ("Model_1", log)
])
pipeline4 = Pipeline([
    ("StandardScaler",std),
    ("Model_2",knn)
])
pipeline5 = Pipeline([
    ("MinMax", minmax),
    ("Model_1",knn)
])
pipeline6 = Pipeline([
    ("RobuseScaler", R_scale),
    ("Model_1", knn)
])
```

```python
cv_score= cross_val_score(pipeline1,X_train,y_train, cv=stf,scoring="accuracy")
cv_score
```

```python
print(f"Mean F1-score: {cv_score.mean() * 100:.2f}%")
print(f"Standard Deviation: {cv_score.std() * 100:.2f}%")
```
